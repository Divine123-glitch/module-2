## Chi-Square Test
<br>

This test whether there is a significant association between categorical variables. It compares observed frequencies with expected frequencies in tables.e.g (Checking whether gender is related to voting preference in a population)

## ANOVA (Analysis of Variance)
<br>
Compares the measure of centre of three or more group, most importantly the mean, to see if at least one is different from the other in a significant manner.e.g (Comparing the test scores of students from different schools). 

## Two-tailed Test
<br>
This test determines if a sample statistic is significantly higher or lower than a hypothesized value. e.g (Testing the potency of a new drug than the standard)


## Proportion Test
<br>
This test compares proportions between groups or against a benchmark. e.g (Comparing the proportion of Literate and Illiterate people in two countries)
